```python
# a priori
#   비지도학습
#   연관관계 : A가 나왔을때 B가 나올 확률 -> 추천시스템

# supportData(지지도)
#   0.3으로 세팅 -> 100명중에 30명이상이 구매한것만 치자

# confidenceData(신뢰도)
#   0.7로 세팅 : A -> B가 나올 확률이 0.7이상인것만 보여줘
```

```python
# Scikit-learn에 없음
%pip install apyori
```

```python
trainData = [
                ["치킨", "치킨", "맥주", "소주"],
                ["치킨", "맥주", "소주"],
                ["치킨", "피자", "맥주", "소주"],
                ["치킨", "막걸리"]
            ]
```

```python
from apyori import apriori

result = apriori(trainData, min_support=0.4, min_confidence=0.5)
result = list(result)

for r in result:
    for r2 in r.ordered_statistics:
        print(list(r2.items_base), "를 산 사람이")
        print(list(r2.items_add), "를 살 확률이")
        print(r2.confidence)
        print(r2.lift)
        print("-----")

# frozenset : 수정 불가능한 set
# lift : apyori라이브러리에서 자체 제공
#       (치킨 -> 맥주 살 확률) / (각각 따로 나올 확률)
#       따로 나올 확률보다 같이 나올 확률이 얼마나 되나
```
