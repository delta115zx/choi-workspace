```python
# 네이버 개발자 센터
# 맛집으로 검색한 블로그
# 제목\t내용\t광고여부 txt에
from http.client import HTTPSConnection
from urllib.parse import quote
from json import loads

def removeB(txt):
    return txt.replace("<b>","").replace("</b>","")

q = quote("맛집")
k = {"X-Naver-Client-Id": "Jw3irlagRSw_EoSLFaC6", "X-Naver-Client-Secret": "zDIWqwbkyK"}
hc = HTTPSConnection("openapi.naver.com")
hc.request("GET", "/v1/search/blog.json?query=" + q, headers=k)
resBody = hc.getresponse().read()
hc.close()

f = open("./Users/teacher02/Jan28_1_MachineLearning/naverBlog.txt", "a", encoding="utf-8")
for b in loads(resBody)["items"]:
    title = removeB(b["title"])
    desc = removeB(b["description"])
    print("제목 :", title)
    print("광고 :", desc)
    t = input("광고/정상 : ")
    print("광고여부 :", t)
    f.write("%s\t%s\t%s\n" % (title, desc, t))
f.close()
```
