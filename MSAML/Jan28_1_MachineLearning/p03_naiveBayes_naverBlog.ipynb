```python
f = open("./Users/teacher02/Jan28_1_MachineLearning/naverBlog.txt", "r", encoding="utf-8")

feature = []
label = []
for line in f.readlines():
    line = line.replace("\n","").split("\t")
    feature.append(line[0] + " " + line[1])
    label.append(line[2])

f.close()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

cv = CountVectorizer()
cvResult = cv.fit_transform(feature).toarray()

mnb = MultinomialNB()
mnb.fit(cvResult, label)
```

```python
predictData = input("글 : ")
pdCvResult = cv.transform([predictData])
result = mnb.predict(pdCvResult)[0]
print(result)
```
