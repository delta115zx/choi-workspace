```python
import tensorflow as tf

tf.compat.v1.disable_v2_behavior() # tf 2.x기능 off
```

```python
# 1) 인공신경망 구성
a = tf.constant(10)
b = tf.constant(20)
c = tf.add(a, b)
d = tf.constant("ㅋ")
print(a, c, d) # 실제로 10이 들어감x, 실제로 덧셈x
print("-----")

# 2) 학습데이터를 넣어서 최적의 알고리즘 찾기
s = tf.compat.v1.Session() # 실행세션
# aResult = s.run(a) # 실행세션에 넣어서 돌리면 값 부여
# print(aResult)

dResult = s.run(d)
print(dResult.decode())

cResult = s.run(c) # 연관된거 다 실행됨
print(cResult)

# 3) 완성된 알고리즘 써서 미지의 데이터에 대한 결과 예측
```
