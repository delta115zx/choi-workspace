```python
# linear regression
#   주어진 값들을 최대한 만족시킬만한 a/b를 찾아내서
#   -> 회귀방정식(y = ax + b)을 완성시키고
#   -> 그 식 써서 예측하자

# 인공신경망(DNN) : 다차원 + 다층 회귀분석

# [1, 10], [2, 20], [3, 30] -> [4, ?] : y = 10x     -> 40
# [1, 11], [2, 21], [3, 31] -> [4, ?] : y = 10x + 1 -> 41
# [1, 9], [2, 20], [3, 28.756] -> [4, ?] : y = ax + b -> ??
```

```python
import tensorflow as tf

tf.compat.v1.disable_v2_behavior()
```

```python
# 1) 학습데이터 확보 + 전처리
xData = [1, 2, 3]
yData = [9, 20, 28.756]
```

```python
# 2) 인공신경망 구성
#   변수
#       Variable : a/b : AI가 찾아내야 할 값
#       placeholder : x/y : 학습데이터 들어갈 자리

# 첫 값(0, 1차원, 자료형은 float)
a = tf.Variable(tf.zeros([1], dtype=tf.float64))

# 첫 값(정규분포 따르는 -1 ~ 2 랜덤, 1차원, 자료형은 float)
b = tf.Variable(tf.compat.v1.random_uniform([1], -1, 2, dtype=tf.float64))

x = tf.compat.v1.placeholder(tf.float64)
y = tf.compat.v1.placeholder(tf.float64)

sik = x * a + b
# 처음시작
#       x * 0 + 0.12
#       sik을 통해서 구하면 x=1 -> y=0.12
#       실제 y값은          x=1 -> y=10
#       -> sik을 통해 구한값(0.12), 실제값(10) 차이가 너무 크다 -> a,b를 바꿔보자
# 바꿔서
#       x * 1 + 0.3
#       sik을 통해서 구하면 x=1 -> y=1.3
#       실제 y값은          x=1 -> y=10
#       -> sik을 통해 구한값(1.3), 실제값(10) 차이가 여전히 크다 -> a,b를 바꿔보자
# 반복
#       sik을 통해서 구한 값과 실제 y랑 차이가 별로 없을때까지

# MSE + 차원수 줄이고
loss = tf.reduce_mean(tf.square(y - sik))

# 최적의 a/b를 찾아줄 객체
# o = tf.compat.v1.train.GradientDescentOptimizer(learning_rate=0.1)
o = tf.compat.v1.train.AdamOptimizer(learning_rate=0.1)

# loss를 최소화하는게 목표
goal = o.minimize(loss)
```

```python
# 3) 학습데이터 넣어서 최적의 a/b찾자
s = tf.compat.v1.Session()
s.run(tf.compat.v1.global_variables_initializer()) # Variable들 첫 값 뽑기

for i in range(1000):
    s.run(goal, feed_dict={x:xData, y:yData}) # placeholder x에 xData넣고, ...
    # print(s.run(loss, feed_dict={x:xData, y:yData}))
    #print("y = %fx + %f" % (s.run(a)[0], s.run(b)[0]))

bestA = s.run(a)[0]
bestB = s.run(b)[0]
print(bestA, bestB)
```

```python
# 4) 완성된 알고리즘 써서 미지의 데이터에 대한 결과 예측
newX = float(input("x : "))
newX = [newX]
result = s.run(sik, feed_dict={x:newX})[0]
print(result)
```
