```python
from jamo.jamo import h2j, j2hcj

b = "클"

c = h2j(b) # 조합형 자음/모음으로 분리(VSCode에서 정상적으로 안보임)
print(c)

d = j2hcj(c) # 조합형 자음/모음 -> 온전한 자음/모음으로
print(d)
print(d.find("ㅋ"))
```
