```python
a = ["ㄱ", "ㄴ", "ㄷ"]
b = ["ㄹ", "ㄴ", "ㅁ", "ㄷ"]
```

```python
# b에서 a에 있는것만 콘솔출력
for bb in b:    # in : for문 (문법)
    if bb in a: # in : list에 포함되어있나 (연산자)
        print(bb)
```

```python
# b에 있는거를 c에도(b랑 똑같은 c를)
# c = []
# for bb in b:
#     c.append(bb)
c = [bb for bb in b]
print(c)
```

```python
# b에 있는거중에 a에도 있는것만 d에
# d = []
# for bb in b:
#     if bb in a:
#         d.append(bb)
d = [bb for bb in b if bb in a]
print(d)
```
