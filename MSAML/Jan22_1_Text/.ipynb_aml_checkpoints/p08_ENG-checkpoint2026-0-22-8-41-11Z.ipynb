```python
from nltk.tokenize import word_tokenize
from nltk.tag import pos_tag
from nltk.stem.wordnet import WordNetLemmatizer
from nltk.corpus import wordnet
```

```python
wnl = WordNetLemmatizer()

f = open("./Users/teacher02/Jan22_1_Text/sherlock.txt", "r", encoding="utf-8")

wc = {}
for line in f.readlines():
    line = line.replace("\n","")
    line = word_tokenize(line)
    line = pos_tag(line)
    for word, poomsa in line:
        if poomsa.startswith("V"):
            word = wnl.lemmatize(word, wordnet.VERB)
            if word in wc:
                wc[word] += 1
            else:
                wc[word] = 1
f.close()

word = []
count = []
for k, v in wc.items():
    word.append(k)
    count.append(v)

# f2 = open("./Users/teacher02/Jan22_1_Text/result.csv", "a", encoding="utf-8")

# for k, v in wc.items():
#     f2.write("%s,%d\n" % (k, v))
# f2.close()
# 시각화는 다른 프로그램(Excel이든...)

# 그래프 : matplotlib
```
