```python
# unicode : 컴에 글자를 표현하는 방식 중 하나
# Unicode Transf Format - 8bit
from sys import maxunicode
from unicodedata import category

print(maxunicode) # unicode 문자 전체 개수
print(chr(65)) # 65번째 unicode문자
print(category("A")) # A는 뭐냐(unicode category 구글링)
```

```python
special = []
for i in range(maxunicode):
    # print(chr(i), category(chr(i)))
    if category(chr(i)).startswith("P"):
        special.append(chr(i))
```

```python
# 삼국지, 특수문자는 날리고
f = open("./Users/teacher02/Jan22_1_Text/sam01.txt", "r", encoding="utf-8")

wc = {}
for line in f.readlines():
    for s in special:
        line = line.replace(s, "")
    line = line.replace("\n","").split(" ")
    for word in line:
        if word in wc:
            wc[word] += 1
        else:
            wc[word] = 1
f.close()

for w, c in wc.items():
    print(w, c)
```
