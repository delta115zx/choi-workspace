```python
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
import mlflow

mlflow.start_run()

featuree = np.array([[80, 20], [95, 5], [10, 90], [90, 10], [5, 95]])
labell = np.array(["액션", "액션", "조폭", "액션", "조폭"])

knc = KNeighborsClassifier(3) 
knc.fit(featuree, labell)

fight = 30
yok = 70
z = np.array([[fight, yok]])
result = knc.predict(z)
print(result[0])

mlflow.sklearn.save_model(sk_model=knc, path="./kwonKNNModel")

mlflow.end_run()
```
