```python
import torch
```

```python
W = torch.tensor(5.0, requires_grad=True)
o = torch.optim.SGD([W], lr=0.01)
```

```python
for epoch in range(10):
    sik = 2 * W

    # PyTorch가 Gradient값을 누적
    o.zero_grad() # 누적되고있는 Gradient값 0으로
    sik.backward() # 역전파=연쇄법칙=미분
    print(W.grad.item())
```
