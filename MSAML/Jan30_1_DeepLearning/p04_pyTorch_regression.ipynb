```python
# TensorFlow 1.x : 완전 수작업
# PyTorch

# TensorFlow 2.x : 자동
import torch
```

```python
xData = torch.FloatTensor([[1], [2], [3]])
yData = torch.FloatTensor([[11], [21], [31]])
```

```python
W = torch.zeros(1, requires_grad=True) # 1차원, GD할 대상
b = torch.zeros(1, requires_grad=True)

o = torch.optim.SGD([W, b], lr=0.01) # SGD써서 W,b찾아라, learning_rate=0.01로
```

```python
for epoch in range(1000):
    sik = xData * W + b
    l = torch.mean((sik - yData) ** 2) # MSE

    o.zero_grad() # ?
    l.backward() # 역전파
    o.step() # 다음단계로

    print("y = x * %f + %f" % (W.item(), b.item()))
```
